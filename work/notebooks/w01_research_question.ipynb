## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

# Research Question

## Decision

Which webpages should be reviewed first for content refresh?

## Unit of Analysis

Each row represents one webpage.

## Output

A ranked list of webpages with a refresh priority score and supporting reason codes.

## Action

Content editors can review the highest-ranked pages first, update outdated content, improve SEO, or monitor pages that may be declining.

## Cost of a Wrong Recommendation

A false positive wastes time reviewing pages that do not actually need improvement.

A false negative may cause an important declining page to be ignored, resulting in continued traffic loss.

## Why Machine Learning?

Simple rules cannot capture all relationships between search performance, engagement, freshness, and visibility.

Machine learning can combine multiple observable signals to rank pages more accurately while still supporting human decision making.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Total Pages:", len(df))

print("\nMedian Word Count")
print(df.groupby("trend_direction")["word_count"].median())

print("\nCorrelation between Search Volume and Impressions")
print(df["search_volume"].corr(df["impressions_90d"]))
```

## Observations

The starter dataset contains 30,000 webpages.

The correlation between search volume and impressions is almost zero, showing that keyword search volume alone does not reliably predict page traffic.

Median word counts are very similar across growing and declining pages, suggesting that content length alone is not the main factor affecting performance.

These observations indicate that multiple search and engagement signals should be considered together when prioritizing content refresh opportunities.
