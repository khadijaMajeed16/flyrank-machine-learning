## 1. Unit of analysis + time window

*One row = one what, over which dates? State it, then verify it below.*

One row represents the daily performance of one content item for one client on one report date.

The development window is March 2026, using the `fact_content_daily_performance` March partition.

The observed time window is 2026-03-01 through 2026-03-31, with 9,841,378 rows.

The analysis focuses on content-level search and engagement performance across clients.

## 2. Fields: feature / label / context / excluded

*Sort every field you plan to touch into these four buckets. Excluded needs a why.*

### Features

The initial features are:

- `gsc_impressions` — Google Search Console search visibility.
- `gsc_clicks` — Google Search Console search clicks.
- `ga4_pageviews` — pageviews measured in GA4.
- `sessions_organic` — organic search sessions.
- `scroll_events` — observed engagement through scroll activity.

These are measurable performance signals rather than identifiers.

### Label / outcome proxy

For the leakage experiment, `gsc_clicks` is treated as an outcome proxy and a label-derived version of it is intentionally introduced as a leaked feature.

The leaked feature is not retained in the final feature set.

### Context

- `report_date`
- `client_hash_id`
- `content_hash_id`
- `client_has_gsc`
- `client_has_ga4`
- `gsc_data_available`
- `ga4_data_available`

These fields describe the observation, client/content identity, or whether a measurement source is available.

### Excluded

The identifier fields `client_hash_id` and `content_hash_id` are excluded from the numeric feature set because they identify entities but do not represent measurable performance signals.

The raw date field `report_date` is also excluded from the numeric feature set because it defines the observation window rather than being a performance measurement.

```python
features = con.sql("""
SELECT
    gsc_impressions,
    gsc_sum_position,
    ga4_pageviews,
    sessions_organic,
    scroll_events,
    gsc_clicks AS label
FROM read_parquet('/content/march_2026.parquet')
WHERE gsc_data_available IS TRUE
""").df()

features.head()
```

### Why these features are available at the decision moment

- `gsc_impressions`: available when GSC data is available for the observation.
- `gsc_sum_position`: available when GSC data is available for the observation.
- `ga4_pageviews`: available when GA4 data is available for the observation.
- `sessions_organic`: available when the relevant analytics data is available.
- `scroll_events`: available when engagement tracking is available.

These are treated as observed measurements rather than guaranteed measurements for every row because data-source availability is uneven.

```python
leakage = features.copy()

leakage["leaked_clicks"] = leakage["label"]

print("Rows:", len(leakage))
print("Leaked feature created:", "leaked_clicks" in leakage.columns)
print(
    "Mean absolute error using leaked_clicks:",
    (leakage["label"] - leakage["leaked_clicks"]).abs().mean()
)
honest_features = leakage.drop(columns=["leaked_clicks"])

print(honest_features.columns.tolist())
```

### Leakage experiment

I intentionally created `leaked_clicks` by copying the outcome proxy `gsc_clicks` into a feature column. This produced a mean absolute error of 0 because the feature contained the outcome itself.

This is leakage because the model would effectively be given the answer. The leaked feature is therefore removed from the final feature frame.

The final feature set does not contain `leaked_clicks`.

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

The March 2026 data does not have complete measurement coverage across all rows.

There are 9,841,378 total rows. GSC data is available for 3,611,061 rows, while GA4 data is available for 413,966 rows. Therefore, analyses using these sources are based on the available subset and should not be treated as complete coverage.

The data shows observed associations and performance measurements. It cannot by itself establish that one feature caused a change in performance, so results should be treated as directional decision-support rather than causal evidence.

The March 2026 slice is used for development. It should not be treated as evidence of performance on the later sealed/test period.
