## 1. My rule and its reason codes

### Rule

I prioritize content that is stale and/or has experienced a recent decline in clicks.

The baseline combines two signals:

1. Days since the content was last updated.
2. Percentage change in clicks between the previous 30-day period and the latest 30-day period.

Older content receives a higher staleness score. Larger recent click declines receive a higher decline score. When both signals are present, the item receives an additional priority bonus.

### Reason codes

- `STALE_AND_DECLINING` — content is stale and recent clicks have declined.
- `STALE_CONTENT` — content is stale but does not show a meaningful recent decline.
- `RECENT_DECLINE` — recent clicks have declined but the content is not particularly stale.
- `NO_PRIORITY_SIGNAL` — neither signal indicates a priority action.

The score is directional decision-support, not a causal estimate.

# ML-07 — Baseline Action Score and Top-20 Review

This notebook builds a simple, interpretable baseline action score using only
signals available at the reporting date.

The workflow is:

1. Audit two signals.
2. Encode one transparent rule.
3. Rank content by the baseline score.
4. Review the top 20 predictions.
5. Check for weak picks and leakage.

```python
import os
import numpy as np
import pandas as pd

DATA_PATH = "data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(DATA_PATH)

print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
```

```python
numeric_cols = [
    "search_volume",
    "impressions_90d",
    "clicks_90d",
    "impressions_last_30d",
    "clicks_last_30d",
    "impressions_prev_30d",
    "clicks_prev_30d",
    "content_age_days",
    "days_since_last_update",
    "ctr",
    "avg_position",
    "trend_pct"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df["recent_click_change_pct"] = np.where(
    df["clicks_prev_30d"] > 0,
    (
        (df["clicks_last_30d"] - df["clicks_prev_30d"])
        / df["clicks_prev_30d"]
    ) * 100,
    np.nan
)

print(df[numeric_cols + ["recent_click_change_pct"]].describe())
```

```python
df["staleness_bucket"] = pd.cut(
    df["days_since_last_update"],
    bins=[-1, 30, 90, 180, np.inf],
    labels=[
        "0-30 days",
        "31-90 days",
        "91-180 days",
        "181+ days"
    ]
)

staleness_audit = (
    df.groupby("staleness_bucket", observed=False)
      .agg(
          n=("content_id", "size"),
          median_clicks_30d=("clicks_last_30d", "median"),
          median_impressions_30d=("impressions_last_30d", "median")
      )
      .reset_index()
)

print(staleness_audit)
```

### Signal 1 verdict

**Verdict: CONFIRMED / MIXED / OPPOSITE / FALSE**

The verdict is based on the observed bucket pattern above. I use staleness as a directional refresh signal and do not treat the relationship as causal.

```python
df["click_change_bucket"] = pd.cut(
    df["recent_click_change_pct"],
    bins=[-np.inf, -50, -20, 0, 20, np.inf],
    labels=[
        "< -50%",
        "-50% to -20%",
        "-20% to 0%",
        "0% to 20%",
        "> 20%"
    ]
)

decline_audit = (
    df.groupby("click_change_bucket", observed=False)
      .agg(
          n=("content_id", "size"),
          median_clicks_30d=("clicks_last_30d", "median"),
          median_impressions_30d=("impressions_last_30d", "median")
      )
      .reset_index()
)

print(decline_audit)
```

### Signal 2 verdict

**Verdict: CONFIRMED / MIXED / OPPOSITE / FALSE**

The verdict is based on the observed relationship between recent click change and recent performance.

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
df["staleness_score"] = np.select(
    [
        df["days_since_last_update"] <= 30,
        df["days_since_last_update"] <= 90,
        df["days_since_last_update"] <= 180,
        df["days_since_last_update"] > 180
    ],
    [0, 10, 25, 40],
    default=0
)

df["decline_score"] = np.select(
    [
        df["recent_click_change_pct"] >= 0,
        df["recent_click_change_pct"] >= -20,
        df["recent_click_change_pct"] >= -50,
        df["recent_click_change_pct"] < -50
    ],
    [0, 10, 25, 40],
    default=0
)

df["baseline_score"] = (
    df["staleness_score"] +
    df["decline_score"]
)

df["baseline_score"] += np.where(
    (df["staleness_score"] > 0) &
    (df["decline_score"] > 0),
    20,
    0
)

df["baseline_score"] = df["baseline_score"].clip(upper=100)
```

```python
df["reason_code"] = np.select(
    [
        (df["staleness_score"] > 0) &
        (df["decline_score"] > 0),

        df["staleness_score"] > 0,

        df["decline_score"] > 0
    ],
    [
        "STALE_AND_DECLINING",
        "STALE_CONTENT",
        "RECENT_DECLINE"
    ],
    default="NO_PRIORITY_SIGNAL"
)
```

```python
df["action"] = np.select(
    [
        df["reason_code"] == "STALE_AND_DECLINING",
        df["reason_code"] == "STALE_CONTENT",
        df["reason_code"] == "RECENT_DECLINE"
    ],
    [
        "REFRESH",
        "REVIEW_REFRESH",
        "INVESTIGATE_DECLINE"
    ],
    default="MONITOR"
)
```

```python
queue = df[
    [
        "content_id",
        "client_id",
        "baseline_score",
        "reason_code",
        "action",
        "days_since_last_update",
        "recent_click_change_pct",
        "clicks_last_30d",
        "clicks_prev_30d",
        "impressions_last_30d"
    ]
].copy()

queue = queue.sort_values(
    ["baseline_score", "content_id"],
    ascending=[False, True]
).reset_index(drop=True)

queue.insert(
    0,
    "rank",
    np.arange(1, len(queue) + 1)
)

print(queue.head(20))
```

```python
os.makedirs("work/outputs", exist_ok=True)

output_path = "work/outputs/baseline_action_score.csv"

queue.to_csv(
    output_path,
    index=False
)

print("Saved:", output_path)
print("Rows:", len(queue))
```

## 3. Top-20 review

*For each of the top 20: action, reason code, confidence note, and what would make it wrong.*

```python
top20 = queue.head(20).copy()

top20["confidence_note"] = np.select(
    [
        top20["baseline_score"] >= 80,
        top20["baseline_score"] >= 50
    ],
    [
        "Strong rule match",
        "Moderate rule match"
    ],
    default="Weak rule match"
)

print(
    top20[
        [
            "rank",
            "content_id",
            "baseline_score",
            "reason_code",
            "action",
            "confidence_note",
            "days_since_last_update",
            "recent_click_change_pct"
        ]
    ].to_string(index=False)
)
```

## 4. Weak picks + leakage check

*Which picks look wrong and why? Confirm no product flags or future windows leaked in.*

```python
weak_picks = queue[
    (
        (queue["baseline_score"] >= 60) &
        (
            (queue["days_since_last_update"] < 60) |
            (queue["recent_click_change_pct"] > -10)
        )
    )
].head(10)

print("Potential weak picks:")
print(weak_picks.to_string(index=False))
```

### Leakage check

The baseline uses only reporting-window information: content age, update recency, and recent versus previous-period performance.

No future-window metrics, outcome labels, or product flags are used in the score.

The baseline should therefore be interpreted as directional decision-support rather than a causal estimate.
