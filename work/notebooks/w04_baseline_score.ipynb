# ML-07 — Baseline Action Score and Top-20 Review

This notebook builds a simple, interpretable baseline action score using only
signals available at the reporting date.

The workflow is:

1. Audit two signals.
2. Encode one transparent rule.
3. Rank content by the baseline score.
4. Review the top 20 predictions.
5. Check for weak picks and leakage.

```python
import os
import numpy as np
import pandas as pd

DATA_PATH = "data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(DATA_PATH)

print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
```

```python
df["gsc_impressions"] = pd.to_numeric(
    df["gsc_impressions"], errors="coerce"
)

df["gsc_clicks"] = pd.to_numeric(
    df["gsc_clicks"], errors="coerce"
)

df["gsc_avg_position"] = pd.to_numeric(
    df["gsc_avg_position"], errors="coerce"
)

df["ctr"] = np.where(
    df["gsc_impressions"] > 0,
    df["gsc_clicks"] / df["gsc_impressions"],
    np.nan
)

print(
    df[
        [
            "gsc_impressions",
            "gsc_clicks",
            "gsc_avg_position",
            "ctr"
        ]
    ].describe()
)
```
